```python
import qflib as qf
import numpy as np
import os
import matplotlib.pyplot as plt
plt.style.use('ggplot')

print("qflib version: {0}".format(qf.version()))
print("pid: {0}".format(os.getpid()))
```

```python
spot, T, rd, rf, q = 100.0, 1.0, 0.03, 0.05, 0.01
sigS, sigQ, rho = 0.20, 0.10, 0.30

# discounted quanto forward
qfwd_disc = np.exp(-rd * T) * qf.qFwdPrice(spot, T, rf, q, sigS, sigQ, rho)

for K in [10.0, 1.0, 0.1, 0.001]:
    c = qf.qEuroBS(1, spot, K, T, rd, rf, q, sigS, sigQ, rho)
    print(f"K={K:.3f}  call={c:.6f}  disc_qfwd={qfwd_disc:.6f}  diff={c - qfwd_disc:.2e}")
```
